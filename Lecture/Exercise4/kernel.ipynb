```python
# install scikit-learn
# %pip install scikit-learn

# Bring in numpy(for numerical operations) and matplotlib.pyplot (for graphics)
import numpy as np
import matplotlib.pyplot as plt
from sklearn.kernel_ridge import KernelRidge
```

```python
%matplotlib inline
```

## 1. Data loading and pre-processing

```python
# Load the data
data = np.loadtxt("concentration_intensity.txt")

# Split into columns
concentration = data[:, 0]
intensity = data[:, 1]

print(concentration)
print(intensity)
```

```python
# Reference intensity (zero concentration)
I0 = intensity[concentration == 0][0]

# Compute absorbance
absorbance = -np.log10(intensity / I0)

print(absorbance)
```

```python
# Plot
plt.figure()
plt.scatter(concentration, absorbance)
plt.xlabel("Concentration")
plt.ylabel("Absorbance")
plt.title("Absorbance vs Concentration")
plt.show()
```

## 2. Linear Regression with Zero Intercept

Let's first simplify the problem. Since there is no absorbance at zero concentration, the calibration curve is expected to go through the origin (i.e., the intercept should be 0). The linear model is then $\tilde{y} = \theta x$. In this scenario, the data matrix becomes
$$ \mathbf{X} = 
\begin{pmatrix}
x_1 \\
x_2 \\
\cdots \\
x_m 
\end{pmatrix}
$$

```python
y = absorbance.reshape(-1, 1) # make the label vector
X = concentration.reshape(-1, 1) # make the data matrix

# remove the zero concentration data from our fitting
y = y[1:]
X = X[1:]

print(X)
print(y)

# normal equation
theta = np.linalg.inv(X.T @ X) @ X.T @ y
print("theta is", theta)

# Predicted values at training X
y_pred = X @ theta
print("Predicted values:")
print(y_pred)

# Plot
plt.scatter(X, y, label="Data")
plt.plot(X, y_pred, label="Normal Equation Fit")
plt.xlim(-0.05, 1.05)
plt.ylim(-0.05, 2.55)
plt.legend()
plt.show()
```

## 3. Kernel Regression with Linear Kernel

```python
# Kernel Ridge with linear kernel and no regularization
model = KernelRidge(kernel="linear", alpha=0)
model.fit(X, y)

# Prediction
X_test = np.linspace(0, 5, 100).reshape(-1, 1)
y_pred = model.predict(X_test)

# Predicted values at training X
y_pred = model.predict(X)
print("Predicted values:")
print(y_pred)

# Plot
plt.scatter(X, y, label="Data")
plt.plot(X, y_pred, label="Kernel Regression (Linear, alpha=0)")
plt.xlim(-0.05, 1.05)
plt.ylim(-0.05, 2.55)
plt.legend()
plt.show()
```
