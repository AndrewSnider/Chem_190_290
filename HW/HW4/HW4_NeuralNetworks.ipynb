# Understanding Backpropagation

Backpropagation is the algorithm used to compute **gradients of the loss function with respect to the network parameters** so they can be updated using gradient descent.

Reference:  
http://neuralnetworksanddeeplearning.com/chap2.html

---

# Step 1: Loss Function

We begin with the **Mean Squared Error (MSE)** loss:

$$
C_{MSE} =
\frac{1}{2n}\lVert a^L - y^{GT} \rVert^2
$$

Where:

- $a^L$ = output activations of the final layer  
- $y^{GT}$ = ground truth labels  
- $n$ = number of samples  
- $C$ = cost (loss) function  

The goal of training is to **minimize this cost**.

---

# Step 2: Output Layer Error

The error term for the **final layer** is

$$
\delta^L = \frac{\partial C}{\partial z^L}
= \nabla_a C \odot \sigma'(z^L)
$$

Where:

- $z^L$ = weighted input to the final layer  
- $\sigma'(z^L)$ = derivative of the activation function  
- $\odot$ = element-wise multiplication  

Interpretation:

This equation tells us **how sensitive the loss is to changes in the final layer inputs**.

---

# Step 3: Propagating Error Backwards

For hidden layers, the error is computed recursively:

$$
\delta^l =
\frac{\partial C}{\partial z^l}
=
\left((w^{l+1})^T \delta^{l+1}\right)
\odot
\sigma'(z^l)
$$

Meaning:

1. The error from the **next layer** is propagated backward.  
2. It is multiplied by the **transpose of the weights**.  
3. It is scaled by the **activation derivative**.

This step is the core of **backpropagation**.

---

# Step 4: Gradient of Biases

The gradient with respect to the biases is simply:

$$
\frac{\partial C}{\partial b^l} = \delta^l
$$

Interpretation:

Each neuron’s bias affects the loss directly through the **error term**.

---

# Step 5: Gradient of Weights

The gradient with respect to the weights is

$$
\frac{\partial C}{\partial w^l}
=
(\delta^l)^T a^{l-1}
$$

Where:

- $a^{l-1}$ = activations from the **previous layer**

Interpretation:

The weight update depends on:

- how much **error the neuron has** ($\delta^l$)  
- how strongly the **previous neuron was activated**

---

# Summary of Backpropagation Flow

Forward Pass:

$$
x \rightarrow z^l \rightarrow a^l
$$

Backward Pass:

$$
C \rightarrow \delta^L \rightarrow \delta^{L-1} \rightarrow ... \rightarrow \delta^1
$$

Gradients computed:

- $ \frac{\partial C}{\partial w^l} $
- $ \frac{\partial C}{\partial b^l} $

These gradients are used in **gradient descent** to update parameters.


## Some guidelines for designing a reasonable NN
* The first hidden layer should usually expand the dimension
* The following hidden layers should gradually contract (form funnel shape network)
* Dimension change between layers should not be crazy
* Notice the total number of parameters - ideally less than 1/10 of number of training examples
* Should start from shallow networks, if both training error/ test error are high, then consider increasing network complexity
* Usually suggested to use ReLU activation for hidden layers (can also try tanh)
* Choose the right type of activation for the output layer!
